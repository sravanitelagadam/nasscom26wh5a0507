```python
import seaborn as sns

tips = sns.load_dataset("tips")

tips.head()
```

```python
import matplotlib.pyplot as plt

plt.hist(
tips["total_bill"],
bins=30,
edgecolor="pink"
)
```

```python
counts = tips["smoker"].value_counts()

plt.pie(

counts,

labels=counts.index,

autopct="%1.0f%%",

startangle=0
)
```

```python
sns.scatterplot(

data=tips,

x="tip",

y="total_bill",

hue="smoker"

)
```

```python
sns.barplot(

data=tips,

x="time",

y="total_bill",

hue="smoker"

)
```

```python
sns.boxplot(

data=tips,

x="total_bill",

y="sex"

)
```

```python
sns.countplot(

data=tips,

x="day"

)
```

```python
sns.swarmplot(

data=tips,

x="tip",

y="smoker"

)
```
