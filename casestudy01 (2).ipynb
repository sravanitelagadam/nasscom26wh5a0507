```python
def F(x):
    return 40 + 5*x - 0.2*x**2

def dF(x):
    return 5 - 0.4*x

for x in range(1, 16):
    print("Study hours:", x)
    print("Predicted score:", F(x))
    print("Derivative:", dF(x))
    print()
```

```python


def antiderivative(x):
    return 2.5 * x**2 - (0.2 / 3) * x**3

result = antiderivative(5) - antiderivative(0)

print("Integral from 0 to 5 =", result)
```

```python
import scipy.integrate as integrate

L = lambda x: 5*x - 0.2*x**2

area, error = integrate.quad(L, 0, 5)
print(area)
```

The area represents the total accumulated learning benefit as \(x\) increases from 0 to 5.

```python
import numpy as np

X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values

Y = df["Final_Score"].values

X = np.column_stack((np.ones(len(X)), X))

beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Coefficients:")
print(beta)
print("\nIntercept (b0):", beta[0])
print("Study Hours coefficient (b1):", beta[1])
print("Attendance coefficient (b2):", beta[2])
print("Assignment Score coefficient (b3):", beta[3])
```

```python
import numpy as np

def learning_benefit(x):
    return 10 * x - x**2

x = np.linspace(0, 10, 1000)
y = learning_benefit(x)

area = np.trapezoid(y, x)

print("Area under the learning-benefit curve:", area)
```
