```python
import numpy as np

np.random.seed(42)

flips = np.random.randint(0, 2, size=(100_000, 2))


heads = flips.sum(axis=1)


p_both_heads = np.mean(heads == 2)


p_at_least_one_head = np.mean(heads >= 1)


p_0_heads = np.mean(heads == 0)
p_1_head = np.mean(heads == 1)
p_2_heads = np.mean(heads == 2)

total_probability = p_0_heads + p_1_head + p_2_heads

print("P(both heads):", p_both_heads)
print("P(at least one head):", p_at_least_one_head)

print("\nP(0 heads):", p_0_heads)
print("P(1 head):", p_1_head)
print("P(2 heads):", p_2_heads)

print("\nSum of probabilities:", total_probability)
print("Does it sum to 1?", np.isclose(total_probability, 1))
```

```python
import numpy as np


np.random.seed(42)
rolls = np.random.randint(1, 7, size=100_000)


odd = rolls % 2 == 1
less_than_4 = rolls < 4

p_odd_given_less4 = np.mean(odd[less_than_4])

p_less4 = np.mean(less_than_4)


p_odd = np.mean(odd)

print("P(odd | roll < 4):", p_odd_given_less4)
print("P(roll < 4):", p_less4)
print("P(odd):", p_odd)



if np.isclose(p_odd_given_less4, p_odd, atol=0.01):
    print("The events appear to be independent.")
else:
    print("The events are not independent.")
```

```python

p_spam = 0.20
p_free_given_spam = 0.60
p_free_given_ham  = 0.05


p_ham = 1 - p_spam
p_free = (p_free_given_spam * p_spam) + (p_free_given_ham * p_ham)

print("P('free') =", p_free)


p_spam_given_free = (p_free_given_spam * p_spam) / p_free

print("P(spam | 'free') =", p_spam_given_free)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

samples = np.random.poisson(lam=2, size=10000)

print("Mean of Poisson samples:", samples.mean())


plt.hist(samples, bins=np.arange(samples.min(), samples.max() + 2) - 0.5,
         edgecolor='black')

plt.xlabel("Number of events")
plt.ylabel("Frequency")
plt.title("Poisson Distribution (lambda=2)")
plt.show()


x = np.linspace(-10, 10, 1000)

sigma1 = 1
sigma2 = 2

pdf1 = norm.pdf(x, loc=0, scale=sigma1)
pdf2 = norm.pdf(x, loc=0, scale=sigma2)

plt.plot(x, pdf1, label="sigma = 1")
plt.plot(x, pdf2, label="sigma = 2")

plt.xlabel("x")
plt.ylabel("Probability Density")
plt.title("Gaussian PDFs")
plt.legend()
plt.show()
```

```python
import numpy as np
from scipy.stats import entropy


H4 = entropy([1/4] * 4, base=2)
H6 = entropy([1/6] * 6, base=2)

print("Entropy of 4-sided die:", H4)
print("Entropy of 6-sided die:", H6)



P = np.array([0.7, 0.3])
Q = np.array([0.5, 0.5])

KL = entropy(P, Q, base=2)

print("KL divergence D(P || Q):", KL)

```
