```python

import pandas as pd
import seaborn as sns


df = sns.load_dataset('tips')

numerical_columns = df.select_dtypes(include=['number']).columns.tolist()
categorical_columns = df.select_dtypes(include=['object', 'category']).columns.tolist()

print("Numerical columns:")
print(numerical_columns)

print("\nCategorical columns:")
print(categorical_columns)


print("\nMean of tip:")
print(df['tip'].mean())



print("\nValue counts of sex:")
print(df['sex'].value_counts())
```

```python
import matplotlib.pyplot as plt


mean_tip = df['tip'].mean()
median_tip = df['tip'].median()
mode_tip = df['tip'].mode()

print("Mean:", mean_tip)
print("Median:", median_tip)
print("Mode:", mode_tip.tolist())

if mean_tip > median_tip:
    print("The tip column is right-skewed (mean > median).")
else:
    print("The tip column is not right-skewed based on mean > median.")



plt.figure(figsize=(8, 5))

plt.hist(df['tip'], bins=20, edgecolor='black')

plt.axvline(mean_tip, color='red', linestyle='--',
            label=f'Mean = {mean_tip:.2f}')

plt.axvline(median_tip, color='green', linestyle='-',
            label=f'Median = {median_tip:.2f}')

plt.xlabel('Tip')
plt.ylabel('Frequency')
plt.title('Distribution of Tips')
plt.legend()

plt.show()
```

```python

range_tip = df['tip'].max() - df['tip'].min()
variance_tip = df['tip'].var()
std_tip = df['tip'].std()

print("Range:", range_tip)
print("Variance:", variance_tip)
print("Standard deviation:", std_tip)



Q1 = df['tip'].quantile(0.25)
Q3 = df['tip'].quantile(0.75)
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)


```

```python
import matplotlib.pyplot as plt


five_number_summary = df['tip'].describe()[['min', '25%', '50%', '75%', 'max']]

print("Five-number summary:")
print(five_number_summary)


skewness = df['tip'].skew()

print("\nSkewness:", skewness)


plt.figure(figsize=(8, 4))

plt.boxplot(df['tip'], vert=False)

plt.xlabel('Tip')
plt.title('Box Plot of Tips')

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


correlation_matrix = df.corr(numeric_only=True).round(2)

print("Correlation Matrix:")
print(correlation_matrix)



plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap')
plt.show()



print("\nDescriptive Summary:")
print(df.describe())


```
